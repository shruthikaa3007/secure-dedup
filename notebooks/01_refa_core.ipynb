# 01 REFA Core
Minimal walkthrough of fixed-size chunking plus REFA encryption/decryption through the new `src` package.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import hashlib
from src.crypto.convergent import chunk_file, refa_encrypt, refa_decrypt

data_dir = ROOT / "data"
data_dir.mkdir(exist_ok=True)
sample_path = data_dir / "sample_core.txt"
sample_path.write_text("secure-dedup-bpow core demo\n" * 128, encoding="utf-8")
chunks = chunk_file(str(sample_path), chunk_size=4096)
K_M = hashlib.sha256(b"km-demo").digest()
K_U = hashlib.sha256(b"ku-demo").digest()
ciphertext, token, integrity_tag = refa_encrypt(chunks[0], K_M, K_U)
roundtrip = refa_decrypt(ciphertext, token, K_U)
{"chunks": len(chunks), "first_chunk_roundtrip": roundtrip == chunks[0], "integrity_tag_len": len(integrity_tag)}
```
