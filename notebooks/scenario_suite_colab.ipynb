````python
import json
from pathlib import Path

reports_dir = Path('test_reports')
latest_json = sorted(reports_dir.glob('scenario_suite_report_*.json'))[-1]
latest_md = sorted(reports_dir.glob('scenario_suite_report_*.md'))[-1]
latest_csv = sorted(reports_dir.glob('scenario_suite_metrics_*.csv'))[-1]

print('Latest JSON:', latest_json)
print('Latest MD:', latest_md)
print('Latest CSV:', latest_csv)

report = json.loads(latest_json.read_text())
print('Summary:', report['passed'], '/', report['total_scenarios'], 'passed')
````

````python
import pandas as pd
from pathlib import Path

csv_path = sorted(Path('test_reports').glob('scenario_suite_metrics_*.csv'))[-1]
df = pd.read_csv(csv_path)
df
````

````python
# Optional quick visual: scenario durations
import matplotlib.pyplot as plt

ax = df.plot.bar(x='scenario', y='duration_ms', figsize=(12,4), legend=False)
ax.set_title('Scenario Execution Time (ms)')
ax.set_ylabel('ms')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
````
