````python
import json
from pathlib import Path

reports_dir = Path('test_reports')
latest_json = sorted(reports_dir.glob('deployment_smoke_report_*.json'))[-1]
latest_md = sorted(reports_dir.glob('deployment_smoke_report_*.md'))[-1]

report = json.loads(latest_json.read_text())
print('Latest JSON:', latest_json)
print('Latest MD:', latest_md)
print('Summary:', report['passed'], '/', report['total'], 'passed')
report
````
