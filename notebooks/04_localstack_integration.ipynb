# 04 LocalStack Integration
End-to-end package demo using SecureDedupSystem with LocalStack-backed S3 and DynamoDB, including opaque chunk handles for anti-probing protection.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.system import SecureDedupSystem

data_dir = ROOT / 'data'
data_dir.mkdir(exist_ok=True)
sample_path = data_dir / 'localstack_demo.txt'
sample_path.write_text('localstack integration demo\n' * 256, encoding='utf-8')
system = SecureDedupSystem(ensure_infra=True)
upload_summary = system.upload('user-a', str(sample_path), 'password-a')
downloaded = system.download('user-a', upload_summary['chunk_tags'], 'password-a')
{
    'chunk_count': upload_summary['chunk_count'],
    'chunk_tag_mode': upload_summary['chunk_tag_mode'],
    'privacy_preserving': upload_summary['privacy_preserving'],
    'download_roundtrip': b''.join(downloaded) == sample_path.read_bytes(),
}
```
